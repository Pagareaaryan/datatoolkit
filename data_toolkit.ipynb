Q1. What is NumPy, and why is it widely used in Python?
NumPy, short for Numerical Python, is a core library for numerical and scientific computing in Python. It provides support for working with arrays and matrices, along with a vast collection of mathematical functions to operate on these structures. Because it is implemented in C behind the scenes, NumPy offers faster computation and better memory management compared to native Python data structures, making it widely used for data analysis, machine learning, and scientific applications.

Q2. How does broadcasting work in NumPy?
Broadcasting in NumPy is a technique that allows operations between arrays of different shapes. Instead of replicating data, broadcasting virtually "stretches" smaller arrays across larger ones so that arithmetic operations can be performed without the need for explicit looping or copying, resulting in faster and more memory-efficient calculations.

Q3. What is a Pandas DataFrame?
A Pandas DataFrame is a 2-dimensional, tabular data structure with labeled rows and columns. It can store data of different types (such as integers, strings, or floats) in separate columns, making it ideal for handling structured data similar to SQL tables or Excel spreadsheets. DataFrames provide powerful functionality for data manipulation, cleaning, and analysis.

Q4. Explain the use of the groupby() method in Pandas.
The groupby() method in Pandas is used to split data into groups based on one or more columns, allowing users to perform aggregate operations like sum, mean, or count on each group. After applying these operations, the grouped data can be combined back into a summarized form, which is useful for analyzing categorical variables and performing data aggregation.

Q5. Why is Seaborn preferred for statistical visualizations?
Seaborn is preferred for statistical visualizations because it simplifies the process of creating aesthetically pleasing and informative plots. Built on top of Matplotlib, Seaborn has default styles and color palettes that make complex plots easier to generate. It is particularly useful for visualizing statistical relationships, distributions, and categorical data, and it integrates seamlessly with Pandas DataFrames.

Q6. What are the differences between NumPy arrays and Python lists?
NumPy arrays and Python lists differ primarily in performance and functionality. NumPy arrays are specifically designed for numerical operations and offer faster computation and better memory efficiency. Arrays are homogeneous, meaning all elements must be of the same data type, whereas Python lists can store mixed data types. Additionally, NumPy supports vectorized operations, enabling element-wise calculations without explicit loops, unlike Python lists.

Q7. What is a heatmap, and when should it be used?
A heatmap is a graphical representation of data where values are depicted using varying colors. It is especially useful for visualizing two-dimensional data structures like correlation matrices or frequency tables, where it helps identify patterns, relationships, or intensity of values at a glance.

Q8. What does the term “vectorized operation” mean in NumPy?
A vectorized operation in NumPy refers to performing operations on entire arrays without the need for explicit Python loops. This approach leverages low-level optimizations to execute mathematical computations more efficiently, making the code faster and cleaner.

Q9. How does Matplotlib differ from Plotly?
Matplotlib is a library primarily used for creating static, publication-quality plots. It provides extensive control over plot customization but lacks interactivity by default. Plotly, on the other hand, specializes in interactive visualizations that allow features such as zooming, hovering, and dynamic data presentation. Plotly charts are web-friendly and well-suited for dashboards and exploratory data analysis.

Q10. What is the significance of hierarchical indexing in Pandas?
Hierarchical indexing, also known as MultiIndexing, allows Pandas to handle multi-level or multi-dimensional data by using multiple indexes along rows or columns. This enables more flexible data representation and advanced data manipulations, such as reshaping or working with higher-dimensional data in a simpler tabular form.

Q11. What is the role of Seaborn’s pairplot() function?
Seaborn’s pairplot() function is used to visualize pairwise relationships in a dataset. It generates scatter plots for each possible pair of numerical features while displaying distributions of individual variables along the diagonal. This function is particularly helpful for exploratory data analysis, allowing users to identify trends, patterns, and correlations within the dataset.

Q12. What is the purpose of the describe() function in Pandas?
The describe() function in Pandas provides a statistical summary of numerical columns in a DataFrame. It returns important metrics such as count, mean, standard deviation, minimum, maximum, and quartile values, helping to quickly understand the distribution and central tendencies of the data.

Q13. Why is handling missing data important in Pandas?
Handling missing data is crucial in Pandas because missing values can negatively affect analysis, lead to biased results, or cause errors in computations. Pandas provides multiple methods such as isna(), fillna(), and dropna() to identify, replace, or remove missing values, ensuring the dataset remains clean and reliable for analysis.

Q14. What are the benefits of using Plotly for data visualization?
Plotly offers several benefits, including the creation of interactive and dynamic visualizations that can be embedded in web applications. It supports both 2D and 3D plotting and provides built-in functionality for tooltips, zooming, and data exploration, making it ideal for dashboards and engaging data analysis workflows.

Q15. How does NumPy handle multidimensional arrays?
NumPy handles multidimensional arrays using the ndarray structure, where arrays can have one or multiple dimensions defined by their shape attribute. It allows efficient indexing, slicing, reshaping, and operations across different axes, enabling users to perform complex mathematical computations on large datasets with ease.

Q16. What is the role of Bokeh in data visualization?
Bokeh is a Python library designed for creating interactive and real-time visualizations for web browsers. It allows the development of interactive plots and dashboards with features such as zooming, panning, and filtering, making it suitable for web-based data applications that require user interaction.

Q17. Explain the difference between apply() and map() in Pandas.
The map() function is used for applying a transformation to each element in a Pandas Series. In contrast, the apply() function can be used on both Series and DataFrames to apply a given function along rows or columns. While map() focuses on element-wise operations, apply() is more flexible for performing operations across entire rows or columns.

Q18. What are some advanced features of NumPy?
NumPy offers a variety of advanced features, including broadcasting for arithmetic on arrays with different shapes, linear algebra operations, fast Fourier transforms (FFT), random number generation, masked arrays for handling invalid values, and memory-mapped arrays for working with large datasets beyond available RAM.

Q19. How does Pandas simplify time series analysis?
Pandas simplifies time series analysis by providing robust date-time indexing, resampling methods, time shifting, and rolling window operations. These functions make it easier to perform tasks like frequency conversion, calculating moving averages, and handling irregular time sequences, which are essential for analyzing time-dependent data.

Q20. What is the role of a pivot table in Pandas?
A pivot table in Pandas is used to summarize data by re-organizing and aggregating it based on specified columns and values. It helps in transforming and analyzing large datasets by grouping data and applying aggregate functions like sum or mean to create meaningful summaries.

Q21. Why is NumPy’s array slicing faster than Python’s list slicing?
NumPy’s array slicing is faster because it creates views of the original array rather than copying data, significantly reducing memory overhead. Additionally, NumPy utilizes optimized C-based implementations for slicing operations, which enhances computational performance compared to Python’s native list slicing.

Q22. What are some common use cases for Seaborn?
Common use cases for Seaborn include visualizing data distributions using histograms, kernel density plots, and box plots; comparing categorical variables using bar plots or violin plots; exploring relationships between variables with scatter plots and regression plots; and creating correlation heatmaps to identify associations between features.

```python
# Q2. Write a Pandas script to find the mean of a specific column in a DataFrame
import pandas as pd
df = pd.DataFrame({'Name': ['A', 'B', 'C'], 'Marks': [70, 80, 90]})
mean_value = df['Marks'].mean()
print("Mean of Marks column:", mean_value)
```

```python
# Q3. Create a scatter plot using Matplotlib
import matplotlib.pyplot as plt
x = [1, 2, 3, 4]
y = [10, 20, 25, 30]
plt.scatter(x, y)
plt.title("Scatter Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
# Q4. How do you calculate the correlation matrix using Seaborn and visualize it with a heatmap?
import seaborn as sns
import pandas as pd
data = pd.DataFrame({
    'A': [1, 2, 3, 4],
    'B': [4, 3, 2, 1],
    'C': [5, 6, 7, 8]
})
corr = data.corr()
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
# Q5. Generate a bar plot using Plotly
import plotly.express as px
data = {'Fruits': ['Apple', 'Banana', 'Orange'], 'Quantity': [10, 15, 7]}
df = pd.DataFrame(data)
fig = px.bar(df, x='Fruits', y='Quantity', title='Fruit Quantity Bar Plot')
fig.show()
```

```python
# Q6. Create a DataFrame and add a new column based on an existing column
df = pd.DataFrame({'Name': ['A', 'B', 'C'], 'Score': [80, 90, 70]})
df['Status'] = df['Score'].apply(lambda x: 'Pass' if x >= 75 else 'Fail')
print(df)
```
